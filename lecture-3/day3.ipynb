```python
# importing important libraries

import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
my_image = cv2.imread(r"lotus.jpeg")
my_image_gray = cv2.cvtColor(my_image,cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale Image",my_image_gray)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
sobelx_kernel = np.array([
    [-1,0,1],
    [-2,0,2],
    [-1,0,1]
])
sobely_kernel = np.array([
    [1,2,1],
    [0,0,0],
    [-1,-2,-1]
])

sobelx= cv2.filter2D(
    my_image_gray, 
    -1, 
    sobelx_kernel
)

sobely= cv2.filter2D(
    my_image_gray,
    -1, 
    sobely_kernel
)



cv2.imshow("Sobel X",sobelx)
cv2.imshow("Sobel Y",sobely)
cv2.waitKey(5000)
cv2.destroyAllWindows()

```

```python
sobelx = cv2.Sobel(
    my_image_gray,
    cv2.CV_64F,
    1,
    0,
    ksize=3
)
sobely = cv2.Sobel(
    my_image_gray,
    cv2.CV_64F,
    0,
    1,
    ksize=3
)
# print(sobelx.dtype)
cv2.imshow("Sobel X",sobelx)
cv2.imshow("Sobel Y",sobely)
cv2.waitKey(5000)
cv2.destroyAllWindows()

```

```python
sobel_complete = np.sqrt(sobelx**2 + sobely**2)

sobel_complete = cv2.normalize(
    sobel_complete,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)
print(sobelx.dtype)
sobel_complete = sobel_complete.astype(np.uint8)

# print(sobel_complete.dtype)
# print(sobel_complete.min(), sobel_complete.max())

cv2.imshow(
    "Sobel Complete",
    sobel_complete
)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
laplacian_kernel = np.array([
    [0,-1,0],
    [-1,4,-1],
    [0,-1,0]
])

laplacian = cv2.filter2D(
    my_image_gray,
    -1, 
    laplacian_kernel
)

cv2.imshow("Laplacian",laplacian)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
lap = cv2.Laplacian(
    my_image_gray,
    -1
)
cv2.imshow("Laplacian",lap)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
noise = np.random.normal(
    0,
    5,
    my_image_gray.shape
)

noisy_image = noise + my_image_gray
noisy_image = np.clip(
    noisy_image,
    0,
    255
).astype(np.uint8)

cv2.imshow("Noisy Image",noisy_image)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
sobel_noisy = cv2.Sobel(
    noisy_image,
    -1,
    1,
    1
)

cv2.imshow("Sobel on Noisy Image",sobel_noisy)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```
