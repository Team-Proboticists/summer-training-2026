```python
# importing required libraries

import cv2 
import numpy as np
```

```python
# create our first image background

img=np.full((300,500,3),(0,0,0),dtype=np.uint8)
cv2.imshow("Image1",img) # "Image1" is the name of the window opened , img is our pixel matrix defined above
cv2.waitKey(1000)           # The program waits for the set time (milliseconds)
cv2.destroyAllWindows()  # closes all the open windows
```

```python
# drawing a line
'''
cv2.line(
img, # image array
pt1, # start point (x1, y1)
pt2, # end point (x2, y2)
color, # (B, G, R)
thickness=1, # line thickness
)
'''
cv2.line(img,(350,125),(125,460),(255,0,0),3)
cv2.imshow("Image1",img) 
cv2.waitKey(1000)           
cv2.destroyAllWindows()
```

```python
# drawing an ellipse/circle
'''
cv2.ellipse(
img, # image (numpy array)
center, # (x, y)
axes, # (major_axis, minor_axis)
angle, # rotation of ellipse (degrees)
startAngle, # starting angle of arc
endAngle, # ending angle of arc
color, # (B, G, R)
thickness=1, # line thickness, -1 to fill
)
'''
cv2.ellipse(img,(250,250),(100,50),0,0,360,(0,255,0),3)
cv2.imshow("Image1",img) 
cv2.waitKey(1000)           
cv2.destroyAllWindows()
```

```python
# drawing a rectangle
'''
cv2.rectangle(
img, # image array
pt1, # top-left corner (x1, y1)
pt2, # bottom-right corner (x2, y2)
color, # (B, G, R)
thickness=1, # line thickness, -1 to fill
)
'''
cv2.rectangle(img,(150,150),(400,350),(0,0,255),2)
cv2.imshow("Image1",img) 
cv2.waitKey(1000)           
cv2.destroyAllWindows()
```

```python
# reading a local image

my_image=cv2.imread(r"lotus.jpeg")
my_image_gray=cv2.imread(r"lotus.jpeg",cv2.IMREAD_GRAYSCALE)
cv2.imshow("Image: Original",my_image)
cv2.waitKey(5000)
cv2.imshow("Image: Grayscale",my_image_gray)
cv2.waitKey(5000)
cv2.destroyAllWindows()
```

```python
# conversion between colorspaces

my_image_rgb=cv2.cvtColor(my_image, cv2.COLOR_BGR2RGB)
my_image_hsv=cv2.cvtColor(my_image, cv2.COLOR_BGR2HSV)
cv2.imshow("Image",my_image_hsv)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

```python
## THRESHOLDING / MASKING ON A COIN
coin_img =  cv2.imread(r"coin.jpeg")
# plt.imshow(coin_img)
# image size is 360 x 540 
coin_img = cv2.cvtColor(coin_img,cv2.COLOR_BGR2GRAY)
for i in range(360):
    for j in range(540):
        if coin_img[i][j] > 100:
            coin_img[i][j] =255
        else:
            coin_img[i][j]=0

print(coin_img[0][0])
cv2.imshow("Image",coin_img)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

```python
## how to access our camera video feed

import numpy as np
import cv2 as cv

cap = cv.VideoCapture(0)
if not cap.isOpened():
    print("Cannot open camera")
    exit()
while True:
    # Capture frame-by-frame
    ret, frame = cap.read()

    # if frame is read correctly ret is True
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break
    # Our operations on the frame come here
    gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)
    # Display the resulting frame
    cv.imshow('frame', gray)
    if cv.waitKey(1) == ord('q'):
        break

# When everything done, release the capture
cap.release()
cv.destroyAllWindows()
```
